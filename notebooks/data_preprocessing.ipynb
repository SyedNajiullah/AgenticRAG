# Device Manuals Dataset

Preprocessing

```python
df = pd.read_csv('../datasets/data/medical_device_manuals_dataset.csv')

# Function to create a text document for each device
def create_device_document(row):
    """Merge all device information into a single text document for RAG"""
    
    doc = f"""Device Manual: {row['Device_Name']} {row['Model_Number']}

DEVICE INFORMATION:
- Device Name: {row['Device_Name']}
- Model Number: {row['Model_Number']}
- Manufacturer: {row['Manufacturer']}
- Manual Version: {row['Manual_Version']}
- Publication Date: {row['Publication_Date']}

REGULATORY DETAILS:
- Device Class: {row['Device_Class']}
- Regulatory Approval ID: {row['Regulatory_Approval_ID']}

CLINICAL INFORMATION:
- Patient Population: {row['Patient_Population']}
- Indications for Use: {row['Indications_for_Use']}
- Contraindications: {row['Contraindications']}

TECHNICAL SPECIFICATIONS:
- Sterilization Method: {row['Sterilization_Method']}
- Device Lifetime: {row['Device_Lifetime_Years']} years
- Device Weight: {row['Device_Weight_kg']} kg
- Maximum Operating Temperature: {row['Max_Operating_Temperature_C']}°C

SAFETY METRICS:
- Number of Warnings: {row['Number_of_Warnings']}
- Number of Cautions: {row['Number_of_Cautions']}
"""
    return doc

# Create documents for all devices
documents = []
for idx, row in df.iterrows():
    doc = create_device_document(row)
    documents.append({
        'device_name': row['Device_Name'],
        'model_number': row['Model_Number'],
        'text': doc,
        "Length": len(doc)
    })

# Convert to DataFrame
rag_df = pd.DataFrame(documents)
```

```python
import numpy as np

print(f"Average length of the document is: {np.mean(rag_df["Length"])}")
```

```python
del rag_df["Length"]
```

```python
rag_df.to_csv("../datasets/data_preprocessed/device_manuals.csv", index=False)
```
