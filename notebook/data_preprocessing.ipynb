```python
import pandas as pd
```

```python
df = pd.read_csv("../datasets/data/train.csv")
df.head()
```

```python
df["question_answer"]  = "Question: " + df["Question"].astype(str) + ". " + "Type: " + df["qtype"].astype(str) + ". " + "Answer: " + df["Answer"].astype(str) + ". "

del df["qtype"]
del df["Question"]
del df["Answer"]

df.head()
```
